```python
from langgraph.graph import StateGraph, START, END
from dotenv import load_dotenv
from typing import TypedDict
```

```python
load_dotenv()  # Load environment variables from .env file
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
```

```python
class BlogState(TypedDict):
    title : str
    outline : str
    content : str
```

```python
def create_outline(state : BlogState) -> BlogState:

    title = state['title']

    outline_prompt = f"Generate a detailed outline for a blog post with the title: {title}"
    outline = llm.invoke(outline_prompt).content

    state['outline'] = outline
    return state
```

```python
def create_blog(state : BlogState) -> BlogState:
    outline = state['outline']

    blog_prompt = f"Write a detailed blog post based on the following outline: {outline}"
    content = llm.invoke(blog_prompt).content

    state['content'] = content
    return state
```

```python
graph = StateGraph(BlogState)

# add nodes to the graph
graph.add_node("create_outline", create_outline)
graph.add_node("create_blog", create_blog)

# add edges to the graph
graph.add_edge(START, "create_outline")
graph.add_edge("create_outline", "create_blog")
graph.add_edge("create_blog", END)


workflow = graph.compile()
```

```python
initial_state = {
    "title" : "The Future of Artificial Intelligence"}
final_state = workflow.invoke(initial_state)

print(final_state['outline'])
```
